## Creating Numpy Arrays

```python
import numpy as np
```

```python
# creating a 1D array from a list
my_list = [1,2,3,4,5]
print(my_list)
print(type(my_list))

my_array = np.array(my_list)
print(my_array)
print(type(my_array))
```

```python
# Creating a 2D array from a list of lists
my_list_of_lists = [[1,2,3], [4,5,6], [7,8,9]]
my_2d_array = np.array(my_list_of_lists)
print(my_2d_array)
```

```python
# Specifying the data type
my_array_float = np.array(my_list, dtype=np.float64)
print(my_array_float)
print(my_array_float.dtype)
```

## Creating Array  with specific Values

```python
# Creating an array of zeros and ones
zeros_array = np.zeros((3, 4))
ones_array = np.ones((2,3))
print(zeros_array)
print(ones_array)
```

```python
# Creating an array fileed with 7s
full_array = np.full((2,2), 7)
print(full_array)
```

```python
# Creating an empty array (garbage value -> faster)
empty_array = np.empty((2, 2))
print(empty_array)
```

## Creating Arrays with Sequences

```python
# Creating an array with a range of values
arange_array = np.arange(0, 10, 2) # 2 is the step number
print(arange_array)
```

```python
# Creating an array with evenly spaced values
linspace_array = np.linspace(0, 1, 5) # 5 is the no of elements we eant
print(linspace_array)
```

## Creating Arrays with Random Values

```python
# Creating an array with random values
random_array = np.random.rand(2, 3) # Value between 0 and 1
print(random_array)
```

```python
# Creating an array with random integers
random_int_array = np.random.randint(0, 10, (3, 3)) # low, high(exclusive), shape
print(random_int_array)
```

```python
# Creating an array with random samples from the standard normal distribution
random_normal_array = np.random.randn(2,2)
print(random_normal_array)
```

## Array Attributes

```python
# Checking the number of dimensions
my_2d_array = np.array([[1,2,3], [4,5,6]])
print(my_2d_array.ndim)

my_1d_array = np.array([1,2,3])
print(my_1d_array.ndim)
```

```python
# Checking the shape of the array
print(my_2d_array.shape)
print(my_1d_array.shape)
```

```python
# The total number of elements in the array
print(my_2d_array.size)
print(my_1d_array.size)
```

```python
# Checking the data type of the array
print(my_2d_array.dtype)

float_array = np.array([1.0, 2.0, 3.0])
print(float_array.dtype)
```

```python
# CHecking the itemsize of the array
print(my_2d_array.itemsize)

float_array = np.array([1.0, 2.0, 3.0], dtype=np.float32)
print(float_array.itemsize)
```

```python
# Checking the nbytes(bytes consumed by the elements of the array). [itemsize * size]
print(my_2d_array.nbytes)
print(float_array.nbytes)
```
