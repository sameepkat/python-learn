```python
import pandas as pd

column_names = ['label', 'message']
data = pd.read_csv('datasets/SMSSpamCollection.csv',usecols=[0,1], names=column_names, encoding='ISO-8859-1',
)

print(data.head())
```

```python
num_rows, num_cols = data.shape

print(f"Number of rows: {num_rows}")
print(f"Number of columns: {num_cols}")
```

```python
class_counts = data['label'].value_counts()

print(class_counts)

class_percentages = data['label'].value_counts(normalize=True) * 100
print(class_percentages)
```

```python
spam_messages = data[data['label'] == 'spam']['message'].sample(5)
print("Sample Spam Messages:\n", spam_messages)

ham_messages = data[data['label'] == 'ham']['message'].sample(5)
print("Sample Ham Messages:\n", ham_messages)
```

```python
data['message_length'] = data['message'].apply(len)
print("Spam Message Length Statistics:\n", data[data['label'] == 'spam']['message_length'].describe())
print("Ham Message Length Statistics:\n", data[data['label'] == 'ham']['message_length'].describe())
```

## Using `string` module

### Using string library

```python
import string

text = "Hello, world! This is a test."
punctuation = string.punctuation
print(f"Puctuation to remove: {punctuation}")

# Create a translation table to map punctuation characters to None (remove(
remove_punctuation = str.maketrans('', '', punctuation)
text_without_punctuation = text.translate(remove_punctuation)

print(f"Original text: {text}")
print(f"Text without punctuation: {text_without_punctuation}")
```

### Using regex `(re)`

```python
import re

text = "This is a test with numbers 123 and special characters like $#@."

text_cleaned = re.sub(r'[^a-zA-Z0-9\s]', '', text)
text_without_numbers = re.sub(r'\d+', '', text_cleaned)

print(f"Original text: {text}")
print(f"Text without punctuation and special chaacters: {text_cleaned}")
print(f"Text without numbers: {text_without_numbers}")
```
