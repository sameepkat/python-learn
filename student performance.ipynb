# Student Performance Regression Model

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
```

```python
# Load the dataset
df = pd.read_csv('dataset/student/student-por.csv', delimiter=';')
df.head()
```

```python
# Convert G1 and G3 to numeric (if not already)
df["G1"] = pd.to_numeric(df["G1"], errors='coerce')
df["G1"] = pd.to_numeric(df["G1"], errors='coerce')
```

```python
# Create a scatter plot
plt.scatter(df['G1'], df['G3'], color='blue', alpha=0.5)
plt.xlabel("First Period Grade (G1)")
plt.ylabel("Fina; Grade (G3)")
plt.title("Scatter Plot: G1 vs G3")
plt.show()
```

```python
X = df[['G1']] # Independent
y = df[['G3']] # Dependent

model = LinearRegression()
model.fit(X, y)
```

```python
# Print Slope and Intercept
slope = model.coef_[0]
intercept = model.intercept_
print(f"Slope: {slope}")
print(f"Intercept: {intercept}")
```

```python
# Plot the regression line
plt.scatter(df['G1'], df['G3'], color='blue', alpha=0.5)
plt.plot(df['G1'], model.predict(X), color='red', linewidth=2)
plt.xlabel("First period Grade (G1)")
plt.ylabel("Final Grade (G3)")
plt.title("G1 vs G3 with Regression Line")
plt.show()
```
